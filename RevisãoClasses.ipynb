# Revisão: Classes em Python

# **1. O que é uma Classe?**

Classe é como um molde para criar **objetos (instâncias)**.

Define **atributos** (características) e **métodos** (comportamentos).

```python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def apresentar(self):
        print(f"Olá, meu nome é {self.nome} e tenho {self.idade} anos.")
```

# **2. Atributos e Métodos de Instância**

**Atributos de instância**: São criados dentro do __init__ e pertencem a cada objeto.

**Métodos de instância:** Recebem **self** como primeiro parâmetro e atuam sobre o objeto.

**Atividade — Trabalhando com Atributos e Métodos de Instância**

Você foi contratado para desenvolver uma aplicação simples de controle de estoque de produtos para uma loja. Para isso, crie uma classe chamada Produto que deve possuir:

**Atributos de instância**

* **preco**: o preço do produto.

* **estoque**: a quantidade disponível em estoque.

**Métodos de instância**

* vender(quantidade):
   
 * Reduz a quantidade em estoque, se houver unidades suficientes.

 * Caso contrário, deve exibir uma mensagem de estoque insuficiente.

* repor(quantidade):

 * Aumenta a quantidade em estoque.

* mostrar_informacoes():

 * Exibe o nome, o preço formatado e a quantidade atual em estoque do produto.

**Orientações**

* O método __init__ deve ser utilizado para inicializar os atributos de cada novo produto.

* Todos os métodos devem receber self como primeiro parâmetro.

* Teste a classe criando um objeto do tipo Produto e utilizando seus métodos para vender, repor e mostrar informações.

```python
class Produto:
    def __init__(self, nome, preco, estoque):
        # Atributos de instância
        self.nome = nome
        self.preco = preco
        self.estoque = estoque
```

* Não recebem **self** nem **cls**.

* Funcionam como **funções** dentro da classe, mas não acessam dados da **instância ou da classe**.
