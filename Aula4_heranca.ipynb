# ✅ 1. Introdução ao Conceito de Herança

**Herança** é um dos pilares fundamentais da Programação Orientada a Objetos (POO). Ela permite que uma classe herde **atributos e métodos** de outra classe, promovendo o **reuso de código** e facilitando a manutenção de sistemas complexos.

* **Classe Pai (ou Superclasse):** Classe que contém os atributos e métodos comuns.

* **Classe Filha (ou Subclasse):** Classe que herda da classe pai e pode adicionar ou modificar funcionalidades.


**Exemplo da vida real:**

Imagine uma classe Veiculo. Carros, motos e caminhões são todos veículos e, portanto, podem herdar comportamentos e características básicas como "ligar", "desligar", "mover", entre outros.


# 🧱 2. Criando uma Classe Base e uma Subclasse

**Criando Classe Base e Subclasse**

**Exemmplo 1 - Você deve criar duas classes para representar veículos:**

1. Classe base Veiculo

* Deve ter os atributos marca e ano.

* Deve conter os métodos ligar() e desligar(), que imprimem mensagens indicando que o veículo foi ligado ou desligado.

2. **Subclasse Carro**

* Deve herdar da classe Veiculo.

* Deve adicionar um atributo modelo.

* Deve ter um método exclusivo buzinar() que imprime uma mensagem indicando que o carro está buzinando.

```python
# Classe Base
class Veiculo:
    def __init__(self, marca, ano):
        self.marca = marca
        self.ano = ano

    def ligar(self):
        print(f"O veículo {self.marca} {self.ano} está ligado.")

    def desligar(self):
        print(f"O veículo {self.marca} {self.ano} está desligado.")

# Subclasse
class Carro(Veiculo):
    def __init__(self, marca, ano, modelo):
        super().__init__(marca, ano)  # Reaproveitando o construtor da classe base
        self.modelo = modelo

    def buzinar(self):
        print(f"O carro {self.modelo} está buzinando: BIP BIP!")

# Testando
veiculo = Veiculo("Honda", 2010)
veiculo.ligar()
veiculo.desligar()

print()

carro = Carro("Toyota", 2022, "Corolla")
carro.ligar()
carro.buzinar()
carro.desligar()
```

# 🔄 3. Reaproveitando Construtores com super()

O `super()` permite acessar métodos da superclasse, como o `__init__()`.

Quando a subclasse possui atributos adicionais ou deseja reutilizar o comportamento do construtor da superclasse, usamos super():


**🔹 Exemplo 1: Pessoa e Estudante**

**Enunciado:**
1. Crie uma classe Pessoa com os atributos nome e idade. Em seguida, crie uma subclasse Estudante que herda de Pessoa e adiciona o atributo curso. Reaproveite o construtor da superclasse utilizando super().




```python
# Classe base Pessoa
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def apresentar(self):
        print(f"Olá, meu nome é {self.nome} e tenho {self.idade} anos.")

# Subclasse Estudante
class Estudante(Pessoa):
    def __init__(self, nome, idade, curso):
        super().__init__(nome, idade)  # Reaproveitando o construtor da superclasse
        self.curso = curso

    # Sobrescrevendo o método apresentar para mostrar mais informações
    def apresentar(self):
        print(f"Olá, meu nome é {self.nome}, tenho {self.idade} anos e estudo {self.curso}.")

# Testando as classes
pessoa = Pessoa("Carlos", 40)
pessoa.apresentar()

estudante = Estudante("Ana", 20, "Engenharia")
estudante.apresentar()

```

**🔹 Exemplo 2: Funcionário e Programador**

Implemente uma classe Funcionario com os atributos nome e salario. Depois, crie uma subclasse Programador que inclui o atributo linguagem e aproveita o construtor da superclasse com super().

```python
# Classe base Pessoa
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def apresentar(self):
        print(f"Olá, meu nome é {self.nome} e tenho {self.idade} anos.")

# Subclasse Estudante
class Estudante(Pessoa):
    def __init__(self, nome, idade, curso):
        super().__init__(nome, idade)  # Reaproveitando o construtor da superclasse
        self.curso = curso

    def apresentar(self):
        print(f"Olá, meu nome é {self.nome}, tenho {self.idade} anos e estudo {self.curso}.")

# Testando
pessoa = Pessoa("Carlos", 40)
pessoa.apresentar()

estudante = Estudante("Ana", 20, "Engenharia")
estudante.apresentar()
```

**🔹 Exemplo 3: Funcionário e Programador**

**Enunciado:**

Implemente uma classe Funcionario com os atributos nome e salario. Depois, crie uma subclasse Programador que inclui o atributo linguagem e aproveita o construtor da superclasse com super().

```python
# Classe base
class Funcionario:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario

    def exibir(self):
        print(f"{self.nome} recebe R${self.salario:.2f}")

# Subclasse
class Programador(Funcionario):
    def __init__(self, nome, salario, linguagem):
        super().__init__(nome, salario)
        self.linguagem = linguagem

    def exibir(self):
        print(f"{self.nome} recebe R${self.salario:.2f} e programa em {self.linguagem}")

# Testando
f1 = Funcionario("Paula", 4000)
f1.exibir()

p1 = Programador("Lucas", 5000, "Python")
p1.exibir()
```


# 🔁 4. Tipos de Herança em Python

**➤ Herança Simples**

Uma classe herda de apenas uma superclasse.

É possível herdar de mais de uma classe:


```python
class Veiculo():
    def __init__(self, nome, cor):
        self.nome = nome
        self.cor = cor

class Carro(Veiculo):
    pass
```

**➤ Herança Múltipla**

Uma classe herda de duas ou mais superclasses.

📦 Exemplo: Estoque + Preco → Produto
Vamos usar um contexto comum de sistemas: gerenciamento de produtos.

🔸 Objetivo:
* Criar duas classes com responsabilidades separadas:

  * **Estoque**: gerencia a quantidade.

 * **Preco**: gerencia o preço.

* Criar a classe Produto, que herda de ambas.

```python
# Classe para controle de estoque
class Estoque:
    def __init__(self, quantidade):
        self.quantidade = quantidade

    def mostrar_estoque(self):
        print(f"Quantidade em estoque: {self.quantidade}")

# Classe para controle de preço
class Preco:
    def __init__(self, preco):
        self.preco = preco

    def mostrar_preco(self):
        print(f"Preço: R${self.preco:.2f}")

# Produto herda de Estoque e Preco
class Produto(Estoque, Preco):
    def __init__(self, nome, quantidade, preco):
        self.nome = nome
        Estoque.__init__(self, quantidade)
        Preco.__init__(self, preco)

    def mostrar_dados(self):
        print(f"Produto: {self.nome}")
        self.mostrar_estoque()
        self.mostrar_preco()

# Criando um produto
p1 = Produto("Notebook", 10, 3500.00)
p1.mostrar_dados()
```

**➤ Herança Multinível**

Uma subclasse herda de uma subclasse, formando uma cadeia.

🧱 Exemplo: Pessoa → Funcionario → Gerente
Vamos criar uma hierarquia onde:

* Pessoa é a classe base com nome.

* Funcionario herda de Pessoa e adiciona salário.

* Gerente herda de Funcionario e adiciona o setor que gerencia.

```python
# Classe base
class Pessoa:
    def __init__(self, nome):
        self.nome = nome

    def apresentar(self):
        print(f"Olá, meu nome é {self.nome}.")

# Segunda camada na hierarquia
class Funcionario(Pessoa):
    def __init__(self, nome, salario):
        super().__init__(nome)
        self.salario = salario

    def mostrar_salario(self):
        print(f"Salário: R${self.salario:.2f}")

# Terceira camada na hierarquia
class Gerente(Funcionario):
    def __init__(self, nome, salario, setor):
        super().__init__(nome, salario)
        self.setor = setor

    def mostrar_setor(self):
        print(f"Gerencia o setor de {self.setor}.")

# Criando um gerente
g1 = Gerente("Ana", 8500.00, "TI")
g1.apresentar()
g1.mostrar_salario()
g1.mostrar_setor()

```

# ✍️ 5. Sobrescrita de Métodos (Override)

A sobrescrita ocorre quando a subclasse redefine um método da superclasse com a mesma assinatura.

```python
# Classe base Pessoa
class Pessoa:
    def saudacao(self):
        print("Olá!")  # Método genérico de saudação

# Classe Estudante que herda de Pessoa
class Estudante(Pessoa):
    def saudacao(self):
        # Aqui usamos o método da superclasse para manter a saudação original
        super().saudacao()
        # E acrescentamos uma mensagem mais específica
        print("Olá, sou estudante.")  # Sobrescrita do método

# Instâncias das classes
p = Pessoa()
e = Estudante()

# Chamando os métodos
print("Saudação da Pessoa:")
p.saudacao()  # Saída: Olá!

print("\nSaudação do Estudante:")
e.saudacao()  # Saída: Olá! \n Olá, sou estudante.

```

# 🧪 6. Verificações com isinstance() e issubclass()

* isinstance(objeto, Classe) — Verifica se um objeto é instância de uma determinada classe.

* issubclass(SubClasse, SuperClasse) — Verifica se uma classe é subclasse de outra.

```python
print(isinstance(gato, Animal))       # True
print(issubclass(Cachorro, Animal))     # True
```
